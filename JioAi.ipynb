```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta
import warnings
warnings.filterwarnings('ignore')

print("Libraries loaded ✓")
```

```python
random.seed(42)
np.random.seed(42)

# Sample posts about Jio AI products
positive_posts = [
    "JioBrain is actually impressive, recommended exactly what I needed",
    "Jio's AI customer support resolved my issue in 2 minutes, no human needed",
    "JioCinema AI recommendations are scary good, been watching for 3 hours",
    "Jio AI network optimization is real, my speeds have improved significantly",
    "JioMart AI suggestions are actually relevant unlike Amazon recommendations",
    "Impressed by how Jio is integrating AI into everyday services",
    "JioBrain understands Hindi queries better than Google Assistant honestly",
    "Jio AI chatbot actually solved my billing issue without transferring me",
    "The AI powered Jio app update is smooth and intuitive",
    "Jio is quietly becoming an AI company, this is underrated",
    "JioCinema AI dubbing quality has improved massively this year",
    "Jio AI is making internet accessible to rural India, respect",
]

negative_posts = [
    "JioBrain is just ChatGPT with a Jio logo, nothing original",
    "Jio AI chatbot kept looping me in circles, useless",
    "JioCinema AI recommendations are terrible, shows me same content repeatedly",
    "Jio calling their basic automation AI is a joke",
    "JioMart AI suggested products I already bought, zero intelligence",
    "Jio AI privacy policy is vague, not comfortable sharing data",
    "JioBrain gave me wrong information confidently, dangerous",
    "Jio is just rebranding existing AI tools as their own innovation",
    "AI customer support is a cost cutting excuse, bring back humans",
    "JioCinema AI subtitles are hilariously wrong sometimes",
    "Jio AI hype is massive but product is mediocre",
    "Data hungry Jio AI is not something I trust with my information",
]

neutral_posts = [
    "Tried JioBrain today, interesting concept, needs more work",
    "Jio AI integration is okay, nothing groundbreaking yet",
    "JioCinema AI recommendations are hit or miss for me",
    "Waiting to see how Jio AI develops over next year",
    "JioBrain works for basic queries, complex ones not so much",
    "Jio AI is functional but not wow worthy yet",
    "Mixed feelings about Jio pushing AI into everything",
    "JioMart AI is decent, not sure if it's actual AI or just filters",
    "Jio AI announcements sound big but execution is average",
    "Will give JioBrain more time before forming opinion",
]

# Products
products = ['JioBrain', 'JioCinema AI', 'JioMart AI', 'Jio AI Support', 'Jio Network AI']

# Platforms
platforms = ['Twitter', 'Reddit', 'LinkedIn', 'Quora']

# Generate 500 rows
rows = []
start_date = datetime(2024, 1, 1)

for i in range(500):
    sentiment = random.choices(['Positive', 'Negative', 'Neutral'], weights=[0.45, 0.35, 0.20])[0]
    
    if sentiment == 'Positive':
        text = random.choice(positive_posts)
    elif sentiment == 'Negative':
        text = random.choice(negative_posts)
    else:
        text = random.choice(neutral_posts)
    
    date = start_date + timedelta(days=random.randint(0, 480))
    
    rows.append({
        'post_id': i + 1001,
        'date': date.strftime('%Y-%m-%d'),
        'platform': random.choice(platforms),
        'product': random.choice(products),
        'post_text': text,
        'sentiment': sentiment,
        'likes': random.randint(0, 500),
        'shares': random.randint(0, 100),
        'region': random.choice(['Mumbai', 'Delhi', 'Bangalore', 'Chennai', 'Hyderabad', 'Kolkata', 'Pune', 'Ahmedabad'])
    })

df = pd.DataFrame(rows)
df.to_csv('jio_ai_sentiment.csv', index=False)

print("Dataset created ✓")
print("Shape:", df.shape)
print(df.head(5))
```

```python
import matplotlib.pyplot as plt
```

```python
product_sentiment = df.groupby(['product', 'sentiment']).size().unstack(fill_value=0)

product_sentiment.plot(kind='bar', figsize=(12, 6), 
                       color=['#e74c3c', '#f39c12', '#2ecc71'])
plt.title('Sentiment Breakdown by Jio AI Product', fontsize=14)
plt.xlabel('Product')
plt.ylabel('Number of Posts')
plt.xticks(rotation=30)
plt.legend(title='Sentiment')
plt.tight_layout()
plt.savefig('sentiment_by_product.png', dpi=150)
plt.show()
print("Saved ✓")
```

```python
df['date'] = pd.to_datetime(df['date'])
df['month'] = df['date'].dt.to_period('M').astype(str)

monthly_sentiment = df.groupby(['month', 'sentiment']).size().unstack(fill_value=0)

monthly_sentiment.plot(kind='line', figsize=(14, 6), marker='o',
                       color=['#e74c3c', '#f39c12', '#2ecc71'])
plt.title('Jio AI Sentiment Trend Over Time', fontsize=14)
plt.xlabel('Month')
plt.ylabel('Number of Posts')
plt.xticks(rotation=45)
plt.legend(title='Sentiment')
plt.tight_layout()
plt.savefig('sentiment_trend.png', dpi=150)
plt.show()
print("Saved ✓")
```

```python
platform_sentiment = df.groupby(['platform', 'sentiment']).size().unstack(fill_value=0)

platform_sentiment.plot(kind='bar', figsize=(10, 6),
                        color=['#e74c3c', '#f39c12', '#2ecc71'])
plt.title('Sentiment by Platform', fontsize=14)
plt.xlabel('Platform')
plt.ylabel('Number of Posts')
plt.xticks(rotation=0)
plt.legend(title='Sentiment')
plt.tight_layout()
plt.savefig('sentiment_by_platform.png', dpi=150)
plt.show()
print("Saved ✓")
```

```python
region_sentiment = df.groupby(['region', 'sentiment']).size().unstack(fill_value=0)
region_sentiment['positive_pct'] = round(
    region_sentiment['Positive'] * 100 / region_sentiment.sum(axis=1), 2
)
region_sentiment = region_sentiment.sort_values('positive_pct', ascending=False)

plt.figure(figsize=(10, 5))
plt.bar(region_sentiment.index, region_sentiment['positive_pct'], color='#3498db')
plt.title('Positive Sentiment % by Region', fontsize=14)
plt.xlabel('Region')
plt.ylabel('Positive Sentiment (%)')
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig('sentiment_by_region.png', dpi=150)
plt.show()
print("Saved ✓")
```

```python
engagement = df.groupby('sentiment')[['likes', 'shares']].mean().reset_index()

x = range(len(engagement))
width = 0.35

fig, ax = plt.subplots(figsize=(8, 5))
bars1 = ax.bar([i - width/2 for i in x], engagement['likes'], width, 
               label='Avg Likes', color='#3498db')
bars2 = ax.bar([i + width/2 for i in x], engagement['shares'], width, 
               label='Avg Shares', color='#e67e22')

ax.set_title('Avg Engagement by Sentiment', fontsize=14)
ax.set_xlabel('Sentiment')
ax.set_ylabel('Average Count')
ax.set_xticks(x)
ax.set_xticklabels(engagement['sentiment'])
ax.legend()
plt.tight_layout()
plt.savefig('engagement_by_sentiment.png', dpi=150)
plt.show()
print("Saved ✓")
```
